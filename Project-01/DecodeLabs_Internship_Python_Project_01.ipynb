This project is a menu-driven To-Do List Application developed in Python. It allows users to add tasks, view tasks, and exit the program while demonstrating the use of variables, lists, loops, conditional statements, user input and exception handling.

```python

                                   # To-Do List

                                    # 1. Variable
                                    # 2. List
                                    # 3. append() method
                                    # 4. input()
                                    # 5. print()
                                    # 6. while loop
                                    # 7. for loop
                                    # 8. if statement
                                    # 9. break
                                    # 10. User Input Handling
                                    # 11. Diplaying Tasks
                                    # 12. Menu System
                                    # 13. Program Exit
```

```python
                                 # ==========================================
                                 # Python Internship Project 01
                                 # DecodeLabs Internship
    # To-Do List Application
    # ==========================================

tasks = []                                                                      # Empty list to store tasks

print("=" * 45)                                                                 # Output
print("        WELCOME TO TO-DO LIST")                                          # Output
print("        APPLICATION")                                                    # Output
print("=" * 45)                                                                 # Output

while True:                                                                     # While loop to run the application

    print("\nChoose an Option")                                                 # Output
    print("1. Add Task")                                                        # Output
    print("2. View Tasks")                                                      # Output
    print("3. Exit")                                                             # Output

    try:                                                                        # Try block to handle user input
        choice = int(input("Enter your choice (1-3): "))

        if choice == 1:                                                         # If statement to check the user choice

            task = input("Enter your task: ")

            if task.strip() == "":                                              # Check if the task is empty
                print("Task cannot be empty.")                                  # Output
                continue                                                        # Continue to the next iteration

            else:                                                               # Else statement to add the task
                tasks.append(task)
                print("Task added successfully.")                                # Output

        elif choice == 2:                                                       # If statement to check the user choice

            print("\n========== YOUR TASKS ==========")                         # Output

            if len(tasks) == 0:                                                 # Check if the list is empty
                print("No tasks available.")

            else:                                                               # Else statement to display the tasks
                for index, task in enumerate(tasks, start=1):                   # For loop to display the tasks
                    print(f"{index}. {task}")                                   # Output

        elif choice == 3:                                                       # If statement to check the user choice

            print("\nThank you for using the To-Do List Application.")          # Output
            print("Project Completed Successfully.")
            break

        else:                                                                   # Else statement to handle invalid input
            print("Please enter a number between 1 and 3.")                     # Output

    except ValueError:                                                          # Except block to handle invalid input
        print("Invalid input. Please enter only numbers.")                      # Output
```
