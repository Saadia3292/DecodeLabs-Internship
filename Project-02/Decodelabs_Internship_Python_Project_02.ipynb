**Description:**

This project is a Python-based Expense Tracker that allows users to enter multiple expense amounts, calculates the running total, validates user input and displays a final expense summary.

**Objective:**

To develop a Python program that accepts multiple expense amounts, accumulates the values and displays the total amount spent.



```python
# ============================================================

# DecodeLabs - Python Programming

# Project 2: Expense Tracker

# ============================================================


def display_header():
    """Display the application heading."""
    print("=" * 50)
    print("                 EXPENSE TRACKER")
    print("=" * 50)


def get_expense():
    """Get and validate an expense amount from the user."""

    while True:                                                                 # Start an infinite loop
        try:
            # Get expense amount from user
            expense = float(input("Enter expense amount: "))                    # Get expense amount from user

            # Check if expense is negative
            if expense < 0:                                                     # If expense is negative
                print("Error: Expense cannot be negative.")
                continue

            return expense

        except ValueError:
            # If input is not a valid number
            print("Invalid input. Please enter a valid number.")


def run_expense_tracker():
    """Run the main expense tracking process."""

    # Initialize total expense
    total = 0.0                                                                 # Initialize total expense

    # Initialize expense count
    expense_count = 0                                                           # Initialize expense count

    # Start an infinite loop
    while True:


        new_expense = get_expense()                                             # Get expense amount from user

        total = total + new_expense                                             # Add the new expense to the running total

        expense_count += 1                                                      # Count the number of expenses

        print(f"Expense added: {new_expense:.2f}")                              # (Output) Display the added expense


        # Ask the user whether to continue

        while True:
            choice = input(                                                     # Get user input
                "Do you want to add another expense? (yes/no): "
            ).strip().lower()

            if choice == "yes":                                                 # If user wants to add more expenses
                break

            elif choice == "no":                                                # If user doesn't want to add more expenses
                return total, expense_count

            else:                                                               # If input is invalid
                print("Please enter yes or no.")


def main():
    """Run the Expense Tracker application."""


    display_header()                                                            # Display application Heading


    # Run the expense tracker
    total, expense_count = run_expense_tracker()                                # Get the total and expense count

    # Display final expense summary
    print("\n" + "=" * 50)
    print("                 EXPENSE SUMMARY")
    print("=" * 50)
    print(f"Number of Expenses : {expense_count}")
    print(f"Total Spent        : {total:.2f}")
    print("=" * 50)


# Run the program
if __name__ == "__main__":
    main()
```
