**Description:**

This Python-based Random Password Generator creates secure passwords of the desired length using letters, numbers, and special characters, with randomized and shuffled characters for better security.

**Objective:**

To develop a Python program that generates strong and secure random passwords based on the user's selected length, while ensuring that each password contains at least one letter, one number and one special character.

```python
# ============================================================
# PROJECT 3: RANDOM PASSWORD GENERATOR
# Python Programming Internship - DecodeLabs
# ============================================================


# IMPORT REQUIRED MODULES

import random
import string


# ------------------------------------------------------------
# FUNCTION: generate_password
# PURPOSE: Generate a random and complex password
# ------------------------------------------------------------

def generate_password(length):

    # DEFINE CHARACTER SETS

    letters = string.ascii_letters
    # Contains uppercase and lowercase English letters

    numbers = string.digits
    # Contains numbers from 0 to 9

    special_characters = "!@#$%^&*"
    # Contains special characters that can be used


    # --------------------------------------------------------
    # COMBINE ALL AVAILABLE CHARACTERS
    # --------------------------------------------------------

    all_characters = letters + numbers + special_characters
    # Combine letters, numbers, and special characters


    # --------------------------------------------------------
    # ENSURE PASSWORD COMPLEXITY
    # --------------------------------------------------------

    # Create the first three characters:
    # 1. One random letter
    # 2. One random number
    # 3. One random special character

    password = [
        random.choice(letters),
        random.choice(numbers),
        random.choice(special_characters)
    ]


    # --------------------------------------------------------
    # GENERATE REMAINING CHARACTERS
    # --------------------------------------------------------

    for _ in range(length - 3):

        password.append(random.choice(all_characters))


    # --------------------------------------------------------
    # SHUFFLE PASSWORD CHARACTERS
    # --------------------------------------------------------

    random.shuffle(password)


    # --------------------------------------------------------
    # CONVERT LIST INTO STRING
    # --------------------------------------------------------

    return "".join(password)


# ============================================================
# MAIN PROGRAM
# ============================================================

# DISPLAY PROGRAM HEADER

print("=" * 55)

print("          RANDOM PASSWORD GENERATOR")

print("          Python Programming - Project 3")

print("=" * 55)


# ------------------------------------------------------------
# DISPLAY PROGRAM INSTRUCTION
# ------------------------------------------------------------

print("\nCreate a strong and secure random password.")

print("Password must contain at least 4 characters.")


# ------------------------------------------------------------
# GET AND VALIDATE PASSWORD LENGTH
# ------------------------------------------------------------

# Repeatedly ask the user for a valid password length

while True:

    # HANDLE USER INPUT ERRORS

    try:

        length = int(input("\nEnter password length: "))


        # CHECK MINIMUM PASSWORD LENGTH

        if length < 4:

            print("Error: Password length must be at least 4.")


        # VALID INPUT

        else:

            break


    # HANDLE INVALID NON-NUMERIC INPUT

    except ValueError:

        print("Error: Please enter a valid whole number.")


# ------------------------------------------------------------
# GENERATE PASSWORD
# ------------------------------------------------------------

password = generate_password(length)


# ------------------------------------------------------------
# DISPLAY GENERATED PASSWORD
# ------------------------------------------------------------

print("\n" + "-" * 55)

print("Generated Password:")

print(password)

print("-" * 55)


# ------------------------------------------------------------
# DISPLAY SUCCESS MESSAGE
# ------------------------------------------------------------

print("\nPassword generated successfully!")

print("Thank you for using the Random Password Generator.")


# ============================================================
# END OF PROJECT 3
# RANDOM PASSWORD GENERATOR
# ============================================================
```
